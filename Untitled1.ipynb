```python
import pandas as pd
import matplotlib.pyplot as plt

from mlxtend.frequent_patterns import apriori, association_rules, fpgrowth

from sklearn.model_selection import cross_val_predict, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    auc
)
```

```python
import pandas as pd

data = {
    "outlook": [
        "sunny", "sunny", "overcast", "rainy",
        "rainy", "rainy", "overcast", "sunny",
        "sunny", "rainy", "sunny", "overcast",
        "overcast", "rainy"
    ],

    "temperature": [
        "hot", "hot", "hot", "mild",
        "cool", "cool", "cool", "mild",
        "cool", "mild", "mild", "mild",
        "hot", "mild"
    ],

    "humidity": [
        "high", "high", "high", "high",
        "normal", "normal", "normal", "high",
        "normal", "normal", "normal", "high",
        "normal", "high"
    ],

    "windy": [
        False, True, False, False,
        False, True, True, False,
        False, False, True, True,
        False, True
    ],

    "play": [
        "no", "no", "yes", "yes",
        "yes", "no", "yes", "no",
        "yes", "yes", "yes", "yes",
        "yes", "no"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Convert categorical data into one-hot encoded format

encoded_df = pd.get_dummies(df)

print(encoded_df)
```

```python
from mlxtend.frequent_patterns import apriori, association_rules

frequent_itemsets = apriori(
    encoded_df,
    min_support=0.2,
    use_colnames=True
)

print(frequent_itemsets)
```

```python
# Generate association rules from Apriori frequent itemsets

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.7
)

rules = rules.sort_values("confidence", ascending=False)

print(rules[[
    "antecedents",
    "consequents",
    "support",
    "confidence",
    "lift"
]])
```

```python
# FP-Growth frequent itemsets

from mlxtend.frequent_patterns import fpgrowth

frequent_itemsets_fp = fpgrowth(
    encoded_df,
    min_support=0.2,
    use_colnames=True
)

print(frequent_itemsets_fp)
```

```python
# Generate association rules from FP-Growth

fp_rules = association_rules(
    frequent_itemsets_fp,
    metric="confidence",
    min_threshold=0.7
)

fp_rules = fp_rules.sort_values("confidence", ascending=False)

print(fp_rules[[
    "antecedents",
    "consequents",
    "support",
    "confidence",
    "lift"
]])
```

```python
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data
y = iris.target

print("Features:")
print(X[:5])

print("\nTarget:")
print(y[:5])
```

```python
from sklearn.model_selection import cross_val_predict, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier

# J48-like decision tree
j48 = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

# 10-fold cross validation
cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

y_pred = cross_val_predict(
    j48,
    X,
    y,
    cv=cv
)

print("Predicted values:")
print(y_pred)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    classification_report
)

accuracy = accuracy_score(y, y_pred)

precision = precision_score(
    y,
    y_pred,
    average="weighted"
)

recall = recall_score(
    y,
    y_pred,
    average="weighted"
)

cm = confusion_matrix(y, y_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(y, y_pred))
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.ensemble import RandomForestClassifier

# J48
j48 = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

j48_prob = cross_val_predict(
    j48,
    X,
    y,
    cv=cv,
    method="predict_proba"
)

# Random Forest
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_prob = cross_val_predict(
    rf,
    X,
    y,
    cv=cv,
    method="predict_proba"
)

print("J48 probability shape:", j48_prob.shape)
print("Random Forest probability shape:", rf_prob.shape)
```

```python
# ROC Curves for J48 and Random Forest

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# Load binary classification dataset
data = load_breast_cancer()

X_roc = data.data
y_roc = data.target

# 10-fold cross validation
cv_roc = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

# J48-like model
j48_roc = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

# Random Forest
rf_roc = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Get probabilities
j48_prob = cross_val_predict(
    j48_roc,
    X_roc,
    y_roc,
    cv=cv_roc,
    method="predict_proba"
)[:, 1]

rf_prob = cross_val_predict(
    rf_roc,
    X_roc,
    y_roc,
    cv=cv_roc,
    method="predict_proba"
)[:, 1]

# ROC for J48
fpr_j48, tpr_j48, _ = roc_curve(y_roc, j48_prob)
auc_j48 = auc(fpr_j48, tpr_j48)

# ROC for Random Forest
fpr_rf, tpr_rf, _ = roc_curve(y_roc, rf_prob)
auc_rf = auc(fpr_rf, tpr_rf)

# Plot
plt.figure(figsize=(8, 6))

plt.plot(
    fpr_j48,
    tpr_j48,
    label=f"J48 (AUC = {auc_j48:.3f})"
)

plt.plot(
    fpr_rf,
    tpr_rf,
    label=f"Random Forest (AUC = {auc_rf:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves - J48 vs Random Forest")
plt.legend()
plt.grid()
plt.show()
```

```python
print("========== ROC COMPARISON ==========")

print("J48 AUC          :", round(auc_j48, 3))
print("Random Forest AUC:", round(auc_rf, 3))
```
